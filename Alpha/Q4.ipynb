## A. Reflex Agent for Abnormal Movements in a Crowd

A reflex agent takes the current percept and immediately performs an action. It does not store previous situations. Here, the agent checks crowd movement. If the movement is above a safe limit, it gives an alert. Otherwise, it does nothing. This saves energy because the camera/sensor can stay in a low-power monitoring mode and the agent only takes extra action when needed.


```python


def crowd_reflex_agent(movement):
    safe_limit = 7

    if movement > safe_limit:
        return "Abnormal movement -> Send alert"
    else:
        return "Normal movement -> Continue low power monitoring"


movements = [3, 5, 9, 4, 11]

for m in movements:
    print("Movement:", m, "|", crowd_reflex_agent(m))
```

**How it works:**
- Input/percept = crowd movement value.
- Rule = compare movement with the safe limit.
- Action = alert if movement is abnormal.
- If movement is normal, the agent keeps monitoring without using extra resources.


## B. Utility-Based Agent for Stressed and Tensed Patients

A utility-based agent chooses an action by giving each possible action a utility value. For this example, the agent considers stress reduction, energy usage and water usage. The action with the highest utility is selected. The utilities are only example values for demonstrating the agent.


```python


actions = {
    "Breathing exercise": {"stress": 8, "energy": 1, "water": 0},
    "Short walk": {"stress": 7, "energy": 2, "water": 1},
    "Long exercise": {"stress": 9, "energy": 5, "water": 3},
    "Relaxation audio": {"stress": 6, "energy": 1, "water": 0}
}

def utility(values):

    return values["stress"] - 0.5 * values["energy"] - 0.5 * values["water"]

best_action = None
best_utility = -999

for action, values in actions.items():
    u = utility(values)
    print(action, "=> utility =", round(u, 2))
    if u > best_utility:
        best_utility = u
        best_action = action

print("\nSelected action:", best_action)
```

The utility function gives more importance to reducing stress, but it also gives a penalty to actions that use more energy or water. Therefore, the agent can select a useful action without always selecting the most resource-intensive option.


## C. Goal-Based Agent for a Secure School Van Network

The goal of this agent is to provide a secure commute for students. The agent has a destination and checks possible routes. A route is considered better if it is safe, reasonably short and does not require unnecessary fuel. The goal is not simply the shortest route; security is also considered.


```python


routes = {
    "Route A": {"distance": 10, "safety": 9},
    "Route B": {"distance": 7, "safety": 5},
    "Route C": {"distance": 9, "safety": 8},
    "Route D": {"distance": 12, "safety": 10}
}

required_safety = 8
safe_routes = {}

for route, data in routes.items():
    if data["safety"] >= required_safety:
        safe_routes[route] = data

selected_route = min(safe_routes, key=lambda r: safe_routes[r]["distance"])

print("Safe routes:")
for route, data in safe_routes.items():
    print(route, data)

print("\nSelected route:", selected_route)
```

**Goal:** Safe student commute.

The agent first removes routes that do not meet the safety requirement. Then it chooses the shortest route from the remaining safe routes. This can reduce fuel consumption and therefore CO2 emissions while still keeping security as the main constraint.


## D. Learning Agent to Encourage Sports Activities in Young People

A learning agent can learn from the user's previous activity and gradually improve its recommendations. For example, if a student likes swimming more than running, the agent can increase the probability of recommending swimming. It can also prefer activities that need little water or energy when suitable.


```python


activities = {
    "Walking": {"score": 5, "water": 0, "energy": 1},
    "Cycling": {"score": 5, "water": 0, "energy": 1},
    "Swimming": {"score": 5, "water": 5, "energy": 2},
    "Football": {"score": 5, "water": 1, "energy": 2}
}

def choose_activity(activities):

    values = {}
    for activity, data in activities.items():
        values[activity] = data["score"] - 0.2 * data["water"] - 0.2 * data["energy"]
    return max(values, key=values.get), values

for day in range(1, 4):
    activity, values = choose_activity(activities)
    print("Day", day, "-> Recommended:", activity)


    if activity == "Walking":
        activities[activity]["score"] += 1
    else:
        activities[activity]["score"] += 0.2

print("\nFinal activity scores:")
for activity, data in activities.items():
    print(activity, "=", round(data["score"], 2))
```

## Overall comparison

| Agent | Type | Main input | Main decision | Eco-friendly idea |
|---|---|---|---|---|
| A | Reflex | Crowd movement | Alert or continue monitoring | Low-power monitoring |
| B | Utility-based | Patient condition + resource cost | Highest utility action | Penalizes energy/water use |
| C | Goal-based | Route safety and distance | Safe route with less distance | Less fuel and CO2 |
| D | Learning | Activity history/feedback | Improved activity recommendation | Can prefer lower-resource activities |

These four agents use different decision-making approaches. The reflex agent reacts directly to the current input, the utility agent compares the usefulness of different actions, the goal-based agent works toward a specific goal, and the learning agent improves its decisions using feedback. In all four cases, resource usage can be included in the decision so the agents do not unnecessarily waste energy or water.
